```python
def describe_object(obj, name="dataset", depth=0, max_depth=3):
    """
    Recursively inspect a loaded PyTorch object without assuming its structure.
    """
    indent = "    " * depth

    print(f"{indent}{name}: {type(obj).__name__}")

    if isinstance(obj, torch.Tensor):
        print(f"{indent}  shape: {tuple(obj.shape)}")
        print(f"{indent}  dtype: {obj.dtype}")
        print(f"{indent}  min: {obj.min().item() if obj.numel() else 'N/A'}")
        print(f"{indent}  max: {obj.max().item() if obj.numel() else 'N/A'}")
        return

    if depth >= max_depth:
        return

    if isinstance(obj, dict):
        print(f"{indent}  keys: {list(obj.keys())}")

        for key, value in obj.items():
            describe_object(
                value,
                name=f"[{repr(key)}]",
                depth=depth + 1,
                max_depth=max_depth
            )

    elif isinstance(obj, (list, tuple)):
        print(f"{indent}  length: {len(obj)}")

        for index, value in enumerate(obj):
            describe_object(
                value,
                name=f"[{index}]",
                depth=depth + 1,
                max_depth=max_depth
            )

    elif isinstance(obj, TensorDataset):
        print(f"{indent}  number of samples: {len(obj)}")
        print(f"{indent}  number of tensors: {len(obj.tensors)}")

        for index, tensor in enumerate(obj.tensors):
            describe_object(
                tensor,
                name=f"tensor[{index}]",
                depth=depth + 1,
                max_depth=max_depth
            )

    else:
        # Some dataset-like objects expose useful attributes.
        try:
            print(f"{indent}  length: {len(obj)}")
        except TypeError:
            pass


describe_object(dataset_object)
```

## 1.2 Baseline cVAE Architecture

Implement the baseline conditional Variational Autoencoder using the architecture
and training principles from Tutorial 8.3.

The model contains:

- a conditional encoder that estimates latent mean and log-variance;
- the VAE reparameterisation step;
- a conditional decoder that reconstructs the input image;
- reconstruction and KL-divergence loss terms.

The project requires a **7-dimensional latent representation**.

The Tutorial 8.3 architecture is adapted to the supplied Project 6 image size
of `1 × 20 × 20` while retaining the same overall cVAE design.

### 1.2.1 - Conditional Encoder

```python
# 1.2.1 — Conditional Encoder

class cEncoder(nn.Module):
    """
    Conditional VAE encoder adapted from Tutorial 8.3.

    Input:
        x    : image tensor [batch_size, 1, 20, 20]
        cond : one-hot class condition [batch_size, num_classes]

    Output:
        mu     : latent mean [batch_size, z_dim]
        logvar : latent log-variance [batch_size, z_dim]
    """

    def __init__(self, z_dim, num_classes):
        super(cEncoder, self).__init__()

        self.z_dim = z_dim
        self.num_classes = num_classes

        self.encoder_backbone = nn.Sequential(

            # 1 × 20 × 20 -> 16 × 10 × 10
            nn.Conv2d(
                1,
                16,
                kernel_size=3,
                stride=2,
                padding=1
            ),
            nn.ReLU(),

            # 16 × 10 × 10 -> 32 × 5 × 5
            nn.Conv2d(
                16,
                32,
                kernel_size=3,
                stride=2,
                padding=1
            ),
            nn.ReLU(),

            # 32 × 5 × 5 -> 64 × 3 × 3
            nn.Conv2d(
                32,
                64,
                kernel_size=3,
                stride=2,
                padding=1
            ),
            nn.ReLU(),

            # 64 × 3 × 3 -> 256 × 1 × 1
            nn.Conv2d(
                64,
                256,
                kernel_size=3,
                stride=1,
                padding=0
            ),
            nn.ReLU(),

            nn.Flatten()
        )

        # Concatenate encoded image features with class condition,
        # then predict both mu and logvar.
        self.encoder_head = nn.Sequential(
            nn.Linear(
                256 + num_classes,
                2 * z_dim
            )
        )

    def forward(self, x, cond):

        x = self.encoder_backbone(x)

        # Add the conditional class information.
        x = torch.cat([x, cond], dim=1)

        x = self.encoder_head(x)

        # Split output into latent mean and log-variance.
        mu, logvar = torch.chunk(x, 2, dim=1)

        return mu, logvar
```

### 1.2.2 - Conditionalk Decoder

```python
# 1.2.2 — Conditional Decoder

class cDecoder(nn.Module):
    """
    Conditional VAE decoder adapted from Tutorial 8.3.

    Input:
        z    : latent vector [batch_size, z_dim]
        cond : one-hot class condition [batch_size, num_classes]

    Output:
        x_hat : reconstructed/generated image
                [batch_size, 1, 20, 20]
    """

    def __init__(self, z_dim, num_classes):
        super(cDecoder, self).__init__()

        self.z_dim = z_dim
        self.num_classes = num_classes

        # Combine latent representation and class condition,
        # then project to the decoder feature representation.
        self.decoder_input = nn.Sequential(
            nn.Linear(
                z_dim + num_classes,
                256
            ),
            nn.ReLU()
        )

        self.decoder_backbone = nn.Sequential(

            # 256 -> 256 × 1 × 1
            nn.Unflatten(1, (256, 1, 1)),

            # 256 × 1 × 1 -> 64 × 3 × 3
            nn.ConvTranspose2d(
                256,
                64,
                kernel_size=3,
                stride=1,
                padding=0
            ),
            nn.ReLU(),

            # 64 × 3 × 3 -> 32 × 5 × 5
            nn.ConvTranspose2d(
                64,
                32,
                kernel_size=3,
                stride=2,
                padding=1,
                output_padding=0
            ),
            nn.ReLU(),

            # 32 × 5 × 5 -> 16 × 10 × 10
            nn.ConvTranspose2d(
                32,
                16,
                kernel_size=3,
                stride=2,
                padding=1,
                output_padding=1
            ),
            nn.ReLU(),

            # 16 × 10 × 10 -> 1 × 20 × 20
            nn.ConvTranspose2d(
                16,
                1,
                kernel_size=3,
                stride=2,
                padding=1,
                output_padding=1
            ),

            # Dataset pixels are in [0, 1].
            nn.Sigmoid()
        )

    def forward(self, z, cond):

        # Condition the latent vector on the requested class.
        z = torch.cat([z, cond], dim=1)

        z = self.decoder_input(z)

        x_hat = self.decoder_backbone(z)

        return x_hat
```

### 1.2.3 - Conditional Variational Autoencoder

```python
# 1.2.3 — Conditional Variational Autoencoder

class cVAE(nn.Module):
    """
    Conditional Variational Autoencoder.

    Combines the conditional encoder and decoder and implements
    the reparameterisation trick used during training.
    """

    def __init__(self, z_dim, num_classes):
        super(cVAE, self).__init__()

        self.encoder = cEncoder(
            z_dim=z_dim,
            num_classes=num_classes
        )

        self.decoder = cDecoder(
            z_dim=z_dim,
            num_classes=num_classes
        )

        self.z_dim = z_dim
        self.num_classes = num_classes

    def encode(self, x, cond):
        return self.encoder(x, cond)

    def decode(self, z, cond):
        return self.decoder(z, cond)

    def reparameterise(self, mu, logvar):
        """
        Reparameterisation trick:

            sigma = exp(0.5 * logvar)
            epsilon ~ N(0, I)
            z = mu + sigma * epsilon
        """

        sigma = torch.exp(0.5 * logvar)
        epsilon = torch.randn_like(sigma)

        z = mu + epsilon * sigma

        return z

    def forward(self, x, cond):

        mu, logvar = self.encode(x, cond)

        z = self.reparameterise(
            mu,
            logvar
        )

        x_hat = self.decode(
            z,
            cond
        )

        return x_hat, mu, logvar
```

### 1.2.4 - Baseline cVAE Loss

```python
# 1.2.4 — Baseline cVAE Loss

def compute_cvae_loss(
    x,
    x_hat,
    mu,
    logvar,
    beta=0.1
):
    """
    Compute the baseline cVAE objective:

        total loss
        = reconstruction loss
        + beta * KL divergence

    This follows the loss formulation used in Tutorial 8.3.
    """

    # Reconstruction loss.
    recon_loss = 100 * F.mse_loss(
        x_hat,
        x,
        reduction="mean"
    )

    # KL divergence between the learned latent distribution
    # and the standard normal prior N(0, I).
    kl_loss = torch.mean(
        -0.5 * torch.sum(
            1
            + logvar
            - mu.pow(2)
            - logvar.exp(),
            dim=1
        )
    )

    total_loss = recon_loss + beta * kl_loss

    return total_loss, recon_loss, kl_loss
```

## 1.3 Baseline Training

Train the baseline cVAE on the supplied training dataset using the Project 6
configuration.

Training and test losses are recorded across epochs to assess convergence.
The resulting training configuration will also provide the shared settings used
where possible by the discriminator-loss model in Task 2, ensuring that the
later comparison remains controlled.

### 1.3.1 batch prepare helper

```python
# 1.3.1 — Batch Preparation Helper

def get_batch(batch, device, num_classes=10):
    """
    Move a dataset batch to the selected device and convert
    integer class labels into one-hot conditional vectors.

    Returns:
        x     : image batch
        cond  : one-hot class conditions
    """

    x = batch[0].to(device)

    labels = batch[1].to(device)

    cond = F.one_hot(
        labels,
        num_classes=num_classes
    ).float()

    return x, cond
```

### 1.3.2 - Baseline Training Function

```python
# 1.3.2 — Baseline Training Function

def train_baseline_cvae(
    model,
    train_dataloader,
    test_dataloader,
    num_epochs,
    device,
    beta=0.1
):
    """
    Train the baseline cVAE and record training/test loss history.

    Training uses stochastic latent samples through the
    reparameterisation step.

    Validation follows Tutorial 8.3 by decoding the latent mean (mu)
    directly, providing deterministic reconstructions.
    """

    optimizer = optim.AdamW(
        model.parameters(),
        lr=0.001,
        eps=1e-5,
        fused=False,
        betas=(0.75, 0.99)
    )

    history = {
        "train_total": [],
        "train_recon": [],
        "train_kl": [],
        "test_total": [],
        "test_recon": [],
        "test_kl": []
    }

    for epoch in range(num_epochs):

        # ============================================================
        # Training
        # ============================================================

        model.train()

        train_total = 0.0
        train_recon = 0.0
        train_kl = 0.0

        train_batches = 0

        pbar = tqdm(
            train_dataloader,
            unit="batch",
            desc=f"Epoch {epoch + 1:02d}/{num_epochs} - training",
            colour="GREEN"
        )

        for batch in pbar:

            x, cond = get_batch(
                batch,
                device,
                num_classes=NUM_CLASSES
            )

            optimizer.zero_grad()

            # Full stochastic cVAE forward pass.
            x_hat, mu, logvar = model(
                x,
                cond
            )

            total_loss, recon_loss, kl_loss = compute_cvae_loss(
                x,
                x_hat,
                mu,
                logvar,
                beta=beta
            )

            total_loss.backward()

            optimizer.step()

            # Record scalar values only.
            train_total += total_loss.item()
            train_recon += recon_loss.item()
            train_kl += kl_loss.item()

            train_batches += 1

            pbar.set_postfix(
                loss=f"{total_loss.item():.4f}"
            )

        train_total /= train_batches
        train_recon /= train_batches
        train_kl /= train_batches

        # ============================================================
        # Validation / Test
        # ============================================================

        model.eval()

        test_total = 0.0
        test_recon = 0.0
        test_kl = 0.0

        test_batches = 0

        with torch.no_grad():

            pbar = tqdm(
                test_dataloader,
                unit="batch",
                desc=f"Epoch {epoch + 1:02d}/{num_epochs} - testing",
                colour="CYAN"
            )

            for batch in pbar:

                x, cond = get_batch(
                    batch,
                    device,
                    num_classes=NUM_CLASSES
                )

                # Tutorial 8.3 validation behaviour:
                # use latent mean rather than stochastic sampling.
                mu, logvar = model.encode(
                    x,
                    cond
                )

                x_hat = model.decode(
                    mu,
                    cond
                )

                total_loss, recon_loss, kl_loss = compute_cvae_loss(
                    x,
                    x_hat,
                    mu,
                    logvar,
                    beta=beta
                )

                test_total += total_loss.item()
                test_recon += recon_loss.item()
                test_kl += kl_loss.item()

                test_batches += 1

        test_total /= test_batches
        test_recon /= test_batches
        test_kl /= test_batches

        # ============================================================
        # Store epoch results
        # ============================================================

        history["train_total"].append(train_total)
        history["train_recon"].append(train_recon)
        history["train_kl"].append(train_kl)

        history["test_total"].append(test_total)
        history["test_recon"].append(test_recon)
        history["test_kl"].append(test_kl)

        print(
            f"\nEpoch {epoch + 1:02d}/{num_epochs} | "
            f"Train: {train_total:.4f} | "
            f"Test: {test_total:.4f} | "
            f"Recon: {test_recon:.4f} | "
            f"KL: {test_kl:.4f}"
        )

    return history
```

```python
# 1.3.3 — Train the Baseline cVAE

BASELINE_EPOCHS = 50

baseline_history = train_baseline_cvae(
    model=baseline_model,
    train_dataloader=train_dataloader,
    test_dataloader=test_dataloader,
    num_epochs=BASELINE_EPOCHS,
    device=device,
    beta=BETA
)
```

## 2.2 Conditional Discriminator

Implement a conditional discriminator that predicts whether an image is real
(from the supplied dataset) or generated/reconstructed by the cVAE.

The discriminator receives both:

- the image; and
- the associated digit-class condition.

Image features are extracted by the discriminator network and combined with the
one-hot class condition before producing a binary real/fake prediction.

The discriminator is trained using binary classification loss and provides the
adversarial signal used to extend the cVAE objective.

```python
# 2.2.1 — Conditional Discriminator

class ConditionalDiscriminator(nn.Module):
    """
    Conditional discriminator for the Project 6 modified cVAE.

    Input:
        x    : image tensor [batch_size, 1, 20, 20]
        cond : one-hot class condition [batch_size, num_classes]

    Output:
        logits : real/fake logit per sample [batch_size]
    """

    def __init__(self, num_classes):
        super(ConditionalDiscriminator, self).__init__()

        self.num_classes = num_classes

        # Compact CNN feature extractor for 20 × 20 inputs.
        self.features = nn.Sequential(
            # 1 × 20 × 20 -> 32 × 10 × 10
            nn.Conv2d(1, 32, kernel_size=3, stride=2, padding=1),
            nn.LeakyReLU(0.2, inplace=True),

            # 32 × 10 × 10 -> 64 × 5 × 5
            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1),
            nn.LeakyReLU(0.2, inplace=True),

            # 64 × 5 × 5 -> 128 × 3 × 3
            nn.Conv2d(64, 128, kernel_size=3, stride=2, padding=1),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Flatten()
        )

        # 128 × 3 × 3 = 1152 image features, concatenated with class condition.
        self.classifier = nn.Sequential(
            nn.Linear(1152 + num_classes, 128),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(128, 1)
        )

    def forward(self, x, cond):
        h = self.features(x)
        h = torch.cat([h, cond], dim=1)
        logits = self.classifier(h).squeeze(1)
        return logits
```

## 2.3 Modified cVAE Training Procedure

Modify the Task 1 training procedure so that each batch updates both the
discriminator and the cVAE.

For each training batch:

1. the cVAE produces reconstructed/generated images;
2. the discriminator is trained to distinguish:
   - real dataset images from
   - detached cVAE outputs;
3. the cVAE is then updated using:
   - reconstruction loss;
   - KL-divergence loss; and
   - an adversarial loss encouraging its outputs to be classified as real.

The modified cVAE objective can therefore be represented conceptually as:

\[
L_{modified}
=
L_{reconstruction}
+
\beta L_{KL}
+
\lambda_{adv} L_{adversarial}
\]

where `lambda_adv` controls the contribution of the discriminator-based term.

The discriminator and cVAE are optimised separately so that the discriminator
learns the real/fake classification task while the cVAE learns to produce
outputs that better resemble real images.

### 2.3.1 Loss helpers

```python
# 2.3.1 — Discriminator and adversarial loss helpers

bce_logits = nn.BCEWithLogitsLoss()


def compute_discriminator_loss(discriminator, real_x, fake_x, cond):
    """
    Train the discriminator to classify:
        real images  -> 1
        fake images  -> 0

    fake_x should be detached from the cVAE graph.
    """

    real_logits = discriminator(real_x, cond)
    fake_logits = discriminator(fake_x, cond)

    real_targets = torch.ones_like(real_logits)
    fake_targets = torch.zeros_like(fake_logits)

    real_loss = bce_logits(real_logits, real_targets)
    fake_loss = bce_logits(fake_logits, fake_targets)

    d_loss = real_loss + fake_loss

    return d_loss, real_loss, fake_loss


def compute_adversarial_loss(discriminator, fake_x, cond):
    """
    Non-saturating generator-style adversarial loss.

    Encourages the cVAE outputs to be classified as real (target = 1).
    """

    fake_logits = discriminator(fake_x, cond)
    real_targets = torch.ones_like(fake_logits)

    adv_loss = bce_logits(fake_logits, real_targets)

    return adv_loss


def compute_modified_cvae_loss(
    x,
    x_hat,
    mu,
    logvar,
    discriminator,
    cond,
    beta=0.1,
    lambda_adv=0.1
):
    """
    Modified cVAE objective:

        total = recon + beta * KL + lambda_adv * adversarial
    """

    # Reuse the Task 1 reconstruction + KL terms.
    _, recon_loss, kl_loss = compute_cvae_loss(
        x,
        x_hat,
        mu,
        logvar,
        beta=beta
    )

    adv_loss = compute_adversarial_loss(
        discriminator,
        x_hat,
        cond
    )

    total_loss = recon_loss + beta * kl_loss + lambda_adv * adv_loss

    return total_loss, recon_loss, kl_loss, adv_loss
```

### 2.3.2 - Alternating Discriminator

```python
# 2.3.2 — Alternating discriminator / modified-cVAE training

def train_modified_cvae(
    model,
    discriminator,
    train_dataloader,
    test_dataloader,
    num_epochs,
    device,
    beta=0.1,
    lambda_adv=0.1,
    disc_lr=2e-4,
    disc_betas=(0.5, 0.999)
):
    """
    Train the cVAE with an additional discriminator loss.

    Per batch:
        1. cVAE forward pass
        2. discriminator update on real vs detached fake
        3. cVAE update with recon + KL + adversarial terms
    """

    # Same cVAE optimiser settings as Task 1.
    cvae_optimizer = optim.AdamW(
        model.parameters(),
        lr=0.001,
        eps=1e-5,
        fused=False,
        betas=(0.75, 0.99)
    )

    disc_optimizer = optim.Adam(
        discriminator.parameters(),
        lr=disc_lr,
        betas=disc_betas
    )

    history = {
        "train_total": [],
        "train_recon": [],
        "train_kl": [],
        "train_adv": [],
        "train_d": [],
        "test_total": [],
        "test_recon": [],
        "test_kl": [],
        "test_adv": []
    }

    for epoch in range(num_epochs):

        # ============================================================
        # Training
        # ============================================================

        model.train()
        discriminator.train()

        train_total = 0.0
        train_recon = 0.0
        train_kl = 0.0
        train_adv = 0.0
        train_d = 0.0
        train_batches = 0

        pbar = tqdm(
            train_dataloader,
            unit="batch",
            desc=f"Epoch {epoch + 1:02d}/{num_epochs} - training",
            colour="GREEN"
        )

        for batch in pbar:

            x, cond = get_batch(
                batch,
                device,
                num_classes=NUM_CLASSES
            )

            # --------------------------------------------------------
            # 1. cVAE forward
            # --------------------------------------------------------
            x_hat, mu, logvar = model(x, cond)

            # --------------------------------------------------------
            # 2. Discriminator update (real vs detached fake)
            # --------------------------------------------------------
            for param in discriminator.parameters():
                param.requires_grad = True

            disc_optimizer.zero_grad()

            d_loss, _, _ = compute_discriminator_loss(
                discriminator,
                real_x=x,
                fake_x=x_hat.detach(),
                cond=cond
            )

            d_loss.backward()
            disc_optimizer.step()

            # --------------------------------------------------------
            # 3. cVAE update (recon + KL + adversarial)
            # --------------------------------------------------------
            for param in discriminator.parameters():
                param.requires_grad = False

            cvae_optimizer.zero_grad()

            total_loss, recon_loss, kl_loss, adv_loss = compute_modified_cvae_loss(
                x=x,
                x_hat=x_hat,
                mu=mu,
                logvar=logvar,
                discriminator=discriminator,
                cond=cond,
                beta=beta,
                lambda_adv=lambda_adv
            )

            total_loss.backward()
            cvae_optimizer.step()

            for param in discriminator.parameters():
                param.requires_grad = True

            train_total += total_loss.item()
            train_recon += recon_loss.item()
            train_kl += kl_loss.item()
            train_adv += adv_loss.item()
            train_d += d_loss.item()
            train_batches += 1

            pbar.set_postfix(
                cvae=f"{total_loss.item():.4f}",
                d=f"{d_loss.item():.4f}"
            )

        train_total /= train_batches
        train_recon /= train_batches
        train_kl /= train_batches
        train_adv /= train_batches
        train_d /= train_batches

        # ============================================================
        # Validation / Test (deterministic mu decoding, like Task 1)
        # ============================================================

        model.eval()
        discriminator.eval()

        test_total = 0.0
        test_recon = 0.0
        test_kl = 0.0
        test_adv = 0.0
        test_batches = 0

        with torch.no_grad():

            pbar = tqdm(
                test_dataloader,
                unit="batch",
                desc=f"Epoch {epoch + 1:02d}/{num_epochs} - testing",
                colour="CYAN"
            )

            for batch in pbar:

                x, cond = get_batch(
                    batch,
                    device,
                    num_classes=NUM_CLASSES
                )

                mu, logvar = model.encode(x, cond)
                x_hat = model.decode(mu, cond)

                total_loss, recon_loss, kl_loss, adv_loss = compute_modified_cvae_loss(
                    x=x,
                    x_hat=x_hat,
                    mu=mu,
                    logvar=logvar,
                    discriminator=discriminator,
                    cond=cond,
                    beta=beta,
                    lambda_adv=lambda_adv
                )

                test_total += total_loss.item()
                test_recon += recon_loss.item()
                test_kl += kl_loss.item()
                test_adv += adv_loss.item()
                test_batches += 1

        test_total /= test_batches
        test_recon /= test_batches
        test_kl /= test_batches
        test_adv /= test_batches

        history["train_total"].append(train_total)
        history["train_recon"].append(train_recon)
        history["train_kl"].append(train_kl)
        history["train_adv"].append(train_adv)
        history["train_d"].append(train_d)

        history["test_total"].append(test_total)
        history["test_recon"].append(test_recon)
        history["test_kl"].append(test_kl)
        history["test_adv"].append(test_adv)

        print(
            f"\nEpoch {epoch + 1:02d}/{num_epochs} | "
            f"Train: {train_total:.4f} | "
            f"Test: {test_total:.4f} | "
            f"Recon: {test_recon:.4f} | "
            f"KL: {test_kl:.4f} | "
            f"Adv: {test_adv:.4f} | "
            f"D: {train_d:.4f}"
        )

    return history
```

## 2.4.4,  2.4.5 - Sanity Checks

```python
# 2.4.4 — Modified-model reconstruction sanity check

modified_model.eval()

selected_indices = [
    torch.where(test_labels == digit)[0][0].item()
    for digit in range(NUM_CLASSES)
]

recon_images = test_images[selected_indices].to(device)
recon_labels = test_labels[selected_indices].to(device)

recon_conditions = F.one_hot(
    recon_labels,
    num_classes=NUM_CLASSES
).float()

with torch.no_grad():
    mu, logvar = modified_model.encode(recon_images, recon_conditions)
    reconstructed_images = modified_model.decode(mu, recon_conditions)

originals = recon_images.cpu()
reconstructions = reconstructed_images.cpu()

fig, axes = plt.subplots(2, NUM_CLASSES, figsize=(15, 4))

for digit in range(NUM_CLASSES):
    axes[0, digit].imshow(originals[digit].squeeze(), cmap="gray", vmin=0, vmax=1)
    axes[0, digit].set_title(f"{digit}")
    axes[0, digit].axis("off")

    axes[1, digit].imshow(reconstructions[digit].squeeze(), cmap="gray", vmin=0, vmax=1)
    axes[1, digit].axis("off")

axes[0, 0].set_ylabel("Original", fontsize=11)
axes[1, 0].set_ylabel("Reconstructed", fontsize=11)

plt.suptitle("Modified cVAE — Reconstruction Sanity Check")
plt.tight_layout()
plt.show()
```

```python
# 2.4.5 — Modified-model conditional generation sanity check

modified_model.eval()

SANITY_SAMPLES_PER_CLASS = 5

torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

modified_generated_images = []

with torch.no_grad():
    for digit in range(NUM_CLASSES):
        z = torch.randn(SANITY_SAMPLES_PER_CLASS, Z_DIM, device=device)

        labels = torch.full(
            (SANITY_SAMPLES_PER_CLASS,),
            digit,
            dtype=torch.long,
            device=device
        )

        conditions = F.one_hot(labels, num_classes=NUM_CLASSES).float()

        samples = modified_model.decode(z, conditions)
        modified_generated_images.append(samples.cpu())

modified_generated_images = torch.stack(modified_generated_images)

print(
    "Generated image tensor:",
    tuple(modified_generated_images.shape)
)

fig, axes = plt.subplots(
    NUM_CLASSES,
    SANITY_SAMPLES_PER_CLASS,
    figsize=(8, 14)
)

for digit in range(NUM_CLASSES):
    for sample_idx in range(SANITY_SAMPLES_PER_CLASS):
        axes[digit, sample_idx].imshow(
            modified_generated_images[digit, sample_idx].squeeze(),
            cmap="gray",
            vmin=0,
            vmax=1
        )
        axes[digit, sample_idx].axis("off")

        if sample_idx == 0:
            axes[digit, sample_idx].set_ylabel(f"Class {digit}", fontsize=10)

plt.suptitle(
    "Modified cVAE — Conditional Generation Sanity Check",
    y=0.995
)
plt.tight_layout()
plt.show()
```
